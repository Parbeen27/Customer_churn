```python
missing_count = customer.isnull().sum()
missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage":(
        customer.isnull().mean()*100
    ).round(2)
})
missing_summary = missing_summary.sort_values(
    "missing_percentage",
    ascending=False
)
missing_summary
```

```python
summary = {
    "rows": customer.shape[0],
    "columns": customer.shape[1],
    "duplicated_rows": customer.duplicated().sum(),
    "columns_with_missing_values":customer.isnull().any().sum(),
    "total_missing_values": customer.isnull().sum().sum(),
    "duplicate_columns_name":customer.columns.duplicated().sum()
}
pd.Series(summary,name='value')
```
