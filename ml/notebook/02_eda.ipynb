```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("../../data/Telco_Cusomer_Churn.csv")
```

```python
df.columns.to_list()
```

```python
gender = df.groupby("gender").size()
gen_perc = round(gender / gender.sum() * 100,2)
gen_perc
```

```python
df.head()
```

```python
print(
    f"1. Internet Service: {df["InternetService"].unique().tolist()} \n"
    f"2. contract: {df["Contract"].unique().tolist()} \n"
    f"3. Payment: {df["PaymentMethod"].unique().tolist()}"
)
```

```python
service = df.groupby('InternetService').size().sort_values(ascending=False)
service
```

```python
contracts = df.groupby('Contract').size().sort_values(ascending=False)
contracts
```

```python
payments = df.groupby('PaymentMethod').size().sort_values(ascending=False)
payments
```

```python
plt.Figure(figsize=(8,4))
sns.barplot(x=payments.index,y=payments.values)
plt.title("Payment Methods")
plt.xticks(rotation=45,ha="right")
plt.show()
```
